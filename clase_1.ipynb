```python
#Funcion loops (para secuencias más grandes)
seq_3= 'ACTGGTAGCTGAGATGACGTCGACTGAGTAGGTCTGCATATGCTGAGCTTAGGTATGCTTGTTAGGACGAGCTAGTGGTATGACCGAACGGTAGGTGACGTACTGATCGG'
```

```python
def ContenidoGC2(DNA):
  counter = 0
  for base in DNA:
    if base == 'C' or base =='G':
      counter= counter + 1
  return counter/len(DNA)
```

```python
print(ContenidoGC2(seq_3))
```

```python
from Bio import Entrez
accesion_id = 'HQ930401'# mitochondrion Dryocampa rubicunda (Anisota rubicunda)
handle = Entrez.efetch(
    db='nucleotide',
    id=accesion_id,
    rettype='fasta',
    retmode='text'
)
seq_test=handle.read()
handle.close()
print(seq_test)
```

```python
print(ContenidoGC2(seq_test))
print('El contenido GCpara la secuencia ID', accesion_id, 'es:', ContenidoGC2(seq_test))
print('La secuencia ID pertenece a la mitocondria de Dryocampa rubicunda (Anisota rubicunda)')
```
